```python
import gymnasium as gym

from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.common.monitor import Monitor

# Optional: only if uploading to HuggingFace
from huggingface_sb3 import load_from_hub, package_to_hub
from huggingface_hub import notebook_login
```

```python
import gymnasium as gym

# First, we create our environment called LunarLander-v2
env = gym.make("LunarLander-v2")

# Then we reset this environment
observation, info = env.reset()

for _ in range(20):
    # Take a random action
    action = env.action_space.sample()
    print("Action taken:", action)

    # Do this action in the environment and get
    # next_state, reward, terminated, truncated and info
    observation, reward, terminated, truncated, info = env.step(action)

    # If the game is terminated (in our case we land, crashed) or truncated (timeout)
    if terminated or truncated:
        # Reset the environment
        print("Environment is reset")
        observation, info = env.reset()

env.close()
```

```python
# We create our environment with gym.make("<name_of_the_environment>")
env = gym.make("LunarLander-v2")
env.reset()
print("_____OBSERVATION SPACE_____ \n")
print("Observation Space Shape", env.observation_space.shape)
print("Sample observation", env.observation_space.sample())  # Get a random observation
```

What are these 8 numbers?

They represent the lander's physical status:

X coordinate (horizontal position)

Y coordinate (vertical position)

X velocity (left/right speed)

Y velocity (up/down speed)

Angle (how tilted the lander is)

Angular velocity (how fast angle is changing)

Left leg touching ground? (0 or 1)

Right leg touching ground? (0 or 1)

```python
print("\n _____ACTION SPACE_____ \n")
print("Action Space Shape", env.action_space.n)
print("Action Space Sample", env.action_space.sample())  # Take a random action
```

```python
from stable_baselines3.common.env_util import make_vec_env

# Create the environment
env = make_vec_env("LunarLander-v2", n_envs=16)
```

```python
model = PPO("MlpPolicy", env, verbose=1)
```

```python
from stable_baselines3 import PPO

model = PPO(
    policy="MlpPolicy",
    env=env,
    n_steps=1024,
    batch_size=64,
    n_epochs=4,
    gamma=0.999,
    gae_lambda=0.98,
    ent_coef=0.01,
    verbose=1,
)
```

```python
model.learn(total_timesteps=1000000)
model.save("ppo-LunarLander-v2")
```

```python
model_name='LunarLander-v2'
```

```python
# @title
eval_env = Monitor(gym.make("LunarLander-v2"))
mean_reward, std_reward = evaluate_policy(model, eval_env, n_eval_episodes=10, deterministic=True)
print(f"mean_reward={mean_reward:.2f} +/- {std_reward}")
```
